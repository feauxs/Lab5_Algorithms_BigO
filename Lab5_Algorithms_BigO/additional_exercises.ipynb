```python
# Appendix: Student Inputs Configuration
LAST_NAME = "Campos"
STUDENT_ID = "TUPM-26-12345"  # Replace with your actual Student ID
FAVORITE_ARTIST = "Car Seat Headrest"  # Replace if needed

SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST)
ARTIST_SUM = sum(ord(c.upper()) - 64 for c in FAVORITE_ARTIST if c.isalpha())

print(f"SEED_DIGIT: {SEED_DIGIT}")
print(f"ID_SUM: {ID_SUM}")
print(f"NAME_LENGTH: {NAME_LENGTH}")
print(f"ARTIST_LENGTH: {ARTIST_LENGTH}")
print(f"ARTIST_SUM: {ARTIST_SUM}")
```

```python
# 1. Generate unique original sorted incident records
original_dataset = sorted([(SEED_DIGIT * 7 + i * 13 + ID_SUM) for i in range(15)])

# 2. Determine rotation point using SEED_DIGIT
rotation_index = (SEED_DIGIT % (len(original_dataset) - 2)) + 2
rotated_dataset = original_dataset[rotation_index:] + original_dataset[:rotation_index]

# 3. Target Incident Generation
target_incident = original_dataset[(ID_SUM + ARTIST_SUM) % len(original_dataset)]

# Linear Search Implementation
def linear_search_rotated(data, target):
    comparisons = 0
    for idx, item in enumerate(data):
        comparisons += 1
        if item == target:
            return idx, comparisons
    return -1, comparisons

# Rotated Binary Search Implementation
def binary_search_rotated(data, target):
    low, high = 0, len(data) - 1
    comparisons = 0
    
    while low <= high:
        mid = (low + high) // 2
        comparisons += 1
        if data[mid] == target:
            return mid, comparisons
        
        # Check if left half is sorted
        if data[low] <= data[mid]:
            if data[low] <= target < data[mid]:
                high = mid - 1
            else:
                low = mid + 1
        # Right half must be sorted
        else:
            if data[mid] < target <= data[high]:
                low = mid + 1
            else:
                high = mid - 1
                
    return -1, comparisons

lin_pos, lin_comp = linear_search_rotated(rotated_dataset, target_incident)
rot_pos, rot_comp = binary_search_rotated(rotated_dataset, target_incident)

print("--- Emergency Dispatch Record Search ---")
print("Original Dataset:", original_dataset)
print("Rotated Dataset: ", rotated_dataset)
print("Target Incident: ", target_incident)
print(f"Linear Search -> Position: {lin_pos}, Comparisons: {lin_comp}")
print(f"Rotated Binary Search -> Position: {rot_pos}, Comparisons: {rot_comp}")
```

```python
# 1. Generate sorted datasets A and B
dept_a = sorted([(SEED_DIGIT * 3 + i * 5) for i in range(8)])
dept_b = sorted([(ARTIST_LENGTH * 2 + i * 4 + NAME_LENGTH) for i in range(10)])

# 2. Merge sorted arrays (Two-Pointer Algorithm)
def merge_sorted_records(a, b):
    i = j = 0
    merged = []
    comparisons = 0
    
    while i < len(a) and j < len(b):
        comparisons += 1
        if a[i] <= b[j]:
            merged.append(a[i])
            i += 1
        else:
            merged.append(b[j])
            j += 1
            
    while i < len(a):
        merged.append(a[i])
        i += 1
    while j < len(b):
        merged.append(b[j])
        j += 1
        
    return merged, comparisons

consolidated, comparisons = merge_sorted_records(dept_a, dept_b)
is_sorted = consolidated == sorted(consolidated)

print("--- University Enrollment Records ---")
print("Department A:", dept_a)
print("Department B:", dept_b)
print("Consolidated Dataset:", consolidated)
print("Total Records Before:", len(dept_a) + len(dept_b))
print("Total Records After: ", len(consolidated))
print("Comparison Count:   ", comparisons)
print("Ordering Verified:  ", is_sorted)
```

```python
import random
import time

# 1. Base order dataset generation
base_orders = [(ID_SUM * 5 + i * 11 + SEED_DIGIT) % 500 for i in range(30)]

# 2. Input Conditions
condition_sorted = sorted(base_orders)
condition_reverse = sorted(base_orders, reverse=True)
condition_random = base_orders.copy()

# Algorithms with operation counters
def bubble_sort_perf(data):
    a, comparisons, movements = data.copy(), 0, 0
    for i in range(len(a)):
        swapped = False
        for j in range(len(a) - 1 - i):
            comparisons += 1
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                movements += 3  # standard 3-assignment swap
                swapped = True
        if not swapped:
            break
    return a, comparisons, movements

def selection_sort_perf(data):
    a, comparisons, movements = data.copy(), 0, 0
    for i in range(len(a) - 1):
        minimum = i
        for j in range(i + 1, len(a)):
            comparisons += 1
            if a[j] < a[minimum]:
                minimum = j
        if minimum != i:
            a[i], a[minimum] = a[minimum], a[i]
            movements += 3
    return a, comparisons, movements

def insertion_sort_perf(data):
    a, comparisons, movements = data.copy(), 0, 0
    for i in range(1, len(a)):
        key, j = a[i], i - 1
        movements += 1
        while j >= 0:
            comparisons += 1
            if a[j] <= key:
                break
            a[j + 1] = a[j]
            movements += 1
            j -= 1
        a[j + 1] = key
        movements += 1
    return a, comparisons, movements

# Benchmark Runner
conditions = {
    "Sorted": condition_sorted,
    "Reverse": condition_reverse,
    "Random": condition_random
}

algorithms = [
    ("Bubble Sort", bubble_sort_perf),
    ("Selection Sort", selection_sort_perf),
    ("Insertion Sort", insertion_sort_perf)
]

print(f"{'Condition':<10}{'Algorithm':<16}{'Comparisons':>12}{'Movements':>12}{'Time (s)':>15}")
print("-" * 65)

for cond_name, dataset in conditions.items():
    for alg_name, func in algorithms:
        start = time.perf_counter()
        result, comp, mov = func(dataset)
        elapsed = time.perf_counter() - start
        
        # Verify
        assert result == sorted(dataset), f"{alg_name} failed verification!"
        
        print(f"{cond_name:<10}{alg_name:<16}{comp:>12}{mov:>12}{elapsed:>15.8f}")
```
